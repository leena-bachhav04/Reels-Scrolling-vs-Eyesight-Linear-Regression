```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("reels_eyesight.csv")
```

```python
x=df[["Reels_Scroll_Per_Hour"]]
y=df[["Eye_Sight_Score"]]
x
y
```

```python
plt.scatter(x,y)
plt.xlabel("Reels_Scroll_Per_Hour")
plt.ylabel("Eye_Sight_Score")
plt.title("reels vs eye_sight")
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=42
)

print("X_train:", x_train.shape)
print("X_test:", x_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
x_train
```

```python
y_train
```

```python
y_pred = model.predict(x_test)

print("Predicted:", y_pred)
```

```python
print("Actual:", y_test.values)
print("Predicted:", y_pred)
```

```python
mae = mean_absolute_error(y_test, y_pred)

print("MAE:", mae)
```

```python
mse = mean_squared_error(y_test, y_pred)

print("MSE:", mse)
```

```python
rmse = np.sqrt(mse)

print("RMSE:", rmse)
```

```python
r2 = r2_score(y_test, y_pred)

print("R² Score:", r2)
```

```python
plt.scatter(x, y)

plt.plot(x, model.predict(x))

plt.xlabel("Reels Scroll Per Hour")
plt.ylabel("Eye Sight Score")
plt.title("Reels Scroll vs Eye Sight")

plt.show()
```
